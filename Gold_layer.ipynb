```sql
%python
silver_df = spark.table(
    "medallion_project_dk.default.silver_orders"
)

display(silver_df)
```

```sql
%python
from pyspark.sql.functions import sum as spark_sum

gold_df = silver_df.groupBy("product").agg(
    spark_sum("total_amount").alias("total_sales")
)

display(gold_df)
```

```sql
%python
gold_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(
        "medallion_project_dk.default.gold_product_sales"
    )
```

```sql
%python
gold_table = spark.table(
    "medallion_project_dk.default.gold_product_sales"
)

display(gold_table)

print("Gold Records:", gold_table.count())
```

```sql
%python
from pyspark.sql.functions import col

highest_sales_df = gold_table.orderBy(
    col("total_sales").desc()
)

display(highest_sales_df)
```

```sql
%sql

SELECT *
FROM medallion_project_dk.default.gold_product_sales
ORDER BY total_sales DESC;
```
