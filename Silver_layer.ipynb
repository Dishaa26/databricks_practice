```python
bronze_df = spark.table(
    "medallion_project_dk.default.bronze_orders"
)

display(bronze_df)
```

```python
silver_df = bronze_df.dropDuplicates(["order_id"])

display(silver_df)
```

```python
from pyspark.sql.functions import col

silver_df = silver_df.filter(
    col("price").isNotNull()
)

display(silver_df)
```

```python
silver_df = (
    silver_df
    .withColumn("order_id", col("order_id").cast("int"))
    .withColumn("quantity", col("quantity").cast("int"))
    .withColumn("price", col("price").cast("double"))
)

silver_df.printSchema()
```

```python
silver_df = silver_df.withColumn(
    "total_amount",
    col("quantity") * col("price")
)

display(silver_df)
```

```python
silver_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(
        "medallion_project_dk.default.silver_orders"
    )
```

```python
silver_table = spark.table(
    "medallion_project_dk.default.silver_orders"
)

display(silver_table)

print("Silver records:", silver_table.count())
```
