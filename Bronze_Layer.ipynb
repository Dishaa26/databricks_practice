```python
df = spark.read.csv(
    "/Volumes/medallion_project_dk/default/raw_files/orders.csv",
    header=True,
    inferSchema=True
)

display(df)
```

```python
from pyspark.sql.functions import current_timestamp

bronze_df = df.withColumn(
    "ingestion_time",
    current_timestamp()
)

display(bronze_df)
```

```python
bronze_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(
        "medallion_project_dk.default.bronze_orders"
    )
```

```python
bronze_table = spark.table(
    "medallion_project_dk.default.bronze_orders"
)

display(bronze_table)

print("Total records:", bronze_table.count())
```
